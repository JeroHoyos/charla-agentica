## **Parte 3:** La base de datos

La base es SQLite, que viene con Python: queda en un archivo, `datos/inmobiliaria.db`, sin instalar nada.

```python
import sqlite3
from pathlib import Path

with sqlite3.connect("datos/inmobiliaria.db") as conexion:
    conexion.executescript(Path("datos/inmobiliaria.sql").read_text(encoding="utf-8"))
    for tabla in ["asesores", "apartamentos", "visitas", "politicas"]:
        print(tabla.ljust(14), conexion.execute(f"SELECT count(*) FROM {tabla}").fetchone()[0])
```
